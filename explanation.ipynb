# Examples and Demo

The code takes implements an algorithm to compress a network, to maintain either shortest path lengths, or the 'energy cost' of a path. If the $n$ nodes in the graph have locations $Z_1,...,Z_n$, then we define the energy cost of a path $P$ consisting of points ${p_1,...,p_k}$ as
$$

E(P) := \sum_{i=2}^k \|Z_{i}-Z_{i-1}\|^{\alpha}

$$
for some parameter $\alpha \geq 0$. The case $\alpha = 0$ reduces to the path length. The shortest / cheapest path cost between two nodes in $G$ is
$$

d_G(i,j) = \min_{P : P\text{ is a path between }i\text{ and j}} E(P)

$$
We compress the graph with respect to the distortion measure
$$

    D(G, \hat{G}) := \frac{1}{\binom{n}{2}} \sum_{i<j} \frac{d_{\hat{G}}(i,j)}{d_G(i,j)}

$$




As a high level summary, the algorithm works as follows:

1. Input a graph G, and a parameter $L$.
2. Divide the domain into a hexagonal grid, of side length $L$.
3. For each hexagon, select a 'hub' node, and connect it to all other nodes in the hexagon via its shortest / cheapest path.
4. Connect adjacent hubs via their shortest / cheapest path.

This forms a compressed subgraph $\hat{G}$ of $G$. The algorithm is implemented in the file `subgraph_highways.cpp`.
